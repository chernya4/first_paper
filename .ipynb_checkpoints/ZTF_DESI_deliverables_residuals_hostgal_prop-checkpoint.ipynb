```python
import pandas as pd
import numpy as np
from astropy.io import fits
```

# Step 1 #

Take `ZTF_snia_match_DESI_hostgal_Z_filtered.csv` which has the lightcurve parameters and the DESI `TARGETID`. 

And to that simply add the matching `ZTF_residuals_centered.csv` and only add the `residual_centered` collumn.

Thus combining into a new data file called `ZTF_snia_match_DESI_hostgal_Z_filtered_mu_residuals.csv`. 

```python
# Step 1 — File paths
path_main = "ZTF_DESI_data/ZTF_snia_match_DESI_hostgal_Z_filtered.csv"
path_resid = "ZTF_DESI_data/ZTF_residuals_centered.csv"
output_path = "ZTF_DESI_data/ZTF_snia_match_DESI_hostgal_Z_filtered_mu_residuals.csv"

# Step 2 — Read both datasets
df_main = pd.read_csv(path_main)
df_resid = pd.read_csv(path_resid)

# Step 3 — Merge only the 'residual_centered' column on 'ztfname'
df_merged = df_main.merge(df_resid[['ztfname', 'residual_centered']], on='ztfname', how='left')

# Step 4 — Save to new CSV
df_merged.to_csv(output_path, index=False)

print(f"✅ Combined dataset saved to: {output_path}")
print(f"🧾 Shape: {df_merged.shape}")
```

# Step 2 - Dark #

We are going to take this new .csv file: `ZTF_snia_match_DESI_hostgal_Z_filtered_mu_residuals.csv` and we are going to take the following collumn: `TARGETID` and match host galaxy parameters from the DESI dark iron fits catalog. 



```python
# ---------------------------------------------------
# Paths
# ---------------------------------------------------
ztf_path = "ZTF_DESI_data/ZTF_snia_match_DESI_hostgal_Z_filtered_mu_residuals.csv"

fastspecfit_path = "/global/cfs/cdirs/desi/public/dr1/vac/dr1/fastspecfit/iron/v2.1/catalogs/fastspec-iron-main-dark.fits"

output_path = "ZTF_DESI_data/ZTF_snia_match_DESI_hostgal_Z_filtered_mu_residuals_fastspec_dark.csv"

# ---------------------------------------------------
# 1. Read the ZTF–DESI match file (with residuals)
# ---------------------------------------------------
df_ztf = pd.read_csv(ztf_path)

print("ZTF file shape:", df_ztf.shape)
print("Columns in ZTF file:")
print(df_ztf.columns)

# ---------------------------------------------------
# 2. Read the DESI FastSpecFit dark catalog
# ---------------------------------------------------
with fits.open(fastspecfit_path) as hdul:
    # Assuming the main binary table with galaxy properties is in HDU 1
    data = hdul[1].data
    
    # Convert to a pandas DataFrame and fix endianness
    fastspec_df = pd.DataFrame(np.array(data).byteswap().newbyteorder())

print("FastSpecFit (dark) shape:", fastspec_df.shape)
print("Example columns in FastSpecFit catalog:")
print(fastspec_df.columns[:40])  # just show a subset

# ---------------------------------------------------
# 3. Restrict FastSpecFit to only the TARGETIDs in your ZTF file
# ---------------------------------------------------
targetids = df_ztf["TARGETID"].values
mask = fastspec_df["TARGETID"].isin(targetids)
fastspec_sub = fastspec_df[mask].copy()

print("FastSpecFit rows matching ZTF TARGETIDs:", fastspec_sub.shape[0])

# ---------------------------------------------------
# 4. Choose which host-galaxy parameters to bring over
#    👉 Update this list once you’ve inspected the column names.
# ---------------------------------------------------
host_cols = [
    "TARGETID",
    "VDISP",
    "SFR",
    "LOGMSTAR",
]

# Keep only TARGETID and desired host parameters
fastspec_host = fastspec_sub[host_cols].copy()

# ---------------------------------------------------
# 5. Merge onto the ZTF table by TARGETID
# ---------------------------------------------------
df_merged = df_ztf.merge(fastspec_host, on="TARGETID", how="left")

print("Merged table shape:", df_merged.shape)

# ---------------------------------------------------
# 6. Save to new CSV
# ---------------------------------------------------
df_merged.to_csv(output_path, index=False)
print(f"✅ Saved merged file with FastSpecFit dark host properties to:\n{output_path}")
```

So in the following strategy we will try to get the data that we need by reading TargetID's from the following .fits files:

`desi_galaxy_metadata.fits` and `desi_galaxy_metadata_dark.fits`

And search them for their TARGETID's. This is really good because these .fits files have already been filtered by "Galaxies". 

We will take relevant galactic properties from the ` 📁 HDU 1: FASTSPEC` and save them in another fits file.

This strategy is mainly emloyed because when trying to serach all of `HDU 1: FASTSPEC` from just a .CSV file of `ZTF_snia_match_DESI_hostgal_Z_filtered_mu_residuals.csv` and found that this crashed the kernel. 

So it would be easier to just get a .fits file of ALL releavan desi galaxy metadata, save it, and then use the .csv to search for all these types of things. 



```python
import time
import numpy as np
from astropy.table import Table

# ---------------------------------------------
# Paths
# ---------------------------------------------
meta_path = "DES5YR_DESI_data/desi_galaxy_metadata_dark.fits"
fastspec_path = "/global/cfs/cdirs/desi/public/dr1/vac/dr1/fastspecfit/iron/v2.1/catalogs/fastspec-iron-main-bright.fits"
output_host_path = "ZTF_DESI_data/desi_galaxy_fastspec_hostprops_bright.fits"

start = time.time()

# ---------------------------------------------
# 1. Read the pre-trimmed galaxy metadata file
# ---------------------------------------------
print(f"Reading metadata from: {meta_path}")
meta = Table.read(meta_path, hdu=1)  # your trimmed file has the table in HDU 1

print("Metadata columns:", meta.colnames)
print("Metadata length:", len(meta))

# Ensure TARGETID is integer and get unique list
meta["TARGETID"] = meta["TARGETID"].astype("int64")
targetids = np.unique(meta["TARGETID"])
print("Unique TARGETIDs in metadata:", len(targetids))

# ---------------------------------------------
# 2. Read only the needed columns from FASTSPEC
# ---------------------------------------------
desired_host_cols = [
    "TARGETID",
    "VDISP",
    "SFR",
    "LOGMSTAR",
]

print(f"\nReading FASTSPEC from: {fastspec_path}")
print("Requested columns:", desired_host_cols)

# Read from the FASTSPEC HDU, only these columns
fastspec = Table.read(
    fastspec_path,
    hdu="FASTSPEC",       # HDU name you mentioned
)

# print("FASTSPEC shape (selected cols):", fastspec.shape)
# print("FASTSPEC columns actually read:", fastspec.colnames)

# Make sure TARGETID is int64 to match metadata
fastspec["TARGETID"] = fastspec["TARGETID"].astype("int64")

# ---------------------------------------------
# 3. Filter FASTSPEC rows to those in metadata TARGETIDs
# ---------------------------------------------
mask = np.in1d(fastspec["TARGETID"], targetids)
fastspec_sub = fastspec[mask]

print("\nFASTSPEC rows matching metadata TARGETIDs:", len(fastspec_sub))

# ---------------------------------------------
# 4. Save trimmed host properties to a new FITS
# ---------------------------------------------
fastspec_sub.write(output_host_path, overwrite=True)

end = time.time()
print(f"\n✅ Saved trimmed host-properties table to:\n  {output_host_path}")
print(f"⏳ Total time: {end - start:.2f} seconds")

# ---------------------------------------------
# 5. Quick sanity check of the output file
# ---------------------------------------------
print("\n🔍 Inspecting saved host-properties file...")
host_check = Table.read(output_host_path, hdu=1)
print("Output columns:", host_check.colnames)
print("Output length:", len(host_check))
print("First 5 rows:")
print(host_check[:5])
```

# LOADS DESIRED DATA FROM DESI GALAXY SURVEY #

```python
import time
from astropy.io import fits
from astropy.table import Table

# Paths
fits_path = "/global/cfs/cdirs/desi/public/dr1/vac/dr1/fastspecfit/iron/v2.1/catalogs/fastspec-iron-main-dark.fits"
output_path = "ZTF_DESI_DATA/desi_galaxy_metadata_dark.fits"

start = time.time()

# Load and filter
with fits.open(fits_path, memmap=True) as hdul:
    metadata = Table(hdul["METADATA"].data)
    galaxy_rows = metadata[metadata["SPECTYPE"] == "GALAXY"]
    trimmed = galaxy_rows[["TARGETID", "RA", "DEC", "DESI_TARGET", "SPECTYPE", "Z", "ZWARN", "VDISP", "LOGMSTAR", "DN400", "SFR", "ABSMAG01_SDSS_R", "LOGSSFR", "MSTAR"]]

# Save
trimmed.write(output_path, overwrite=True)

end = time.time()
print(f"\n⏳ Total time: {end - start:.2f} seconds")

# Load and display saved file summary
with fits.open(output_path) as hdul_out:
    print(f"\n📁 Output file: {output_path}")
    print("📋 Columns:", hdul_out[1].columns.names)
    print("🧾 First few rows:")
    print(Table(hdul_out[1].data[:5]))  # display first 5 rows
```

```python
import numpy as np
import pandas as pd
from astropy.io import fits

# --------------------------------------------------------------------
# 1. Load your ZTF–DESI matched SNe hosts CSV
# --------------------------------------------------------------------
csv_path = "/global/homes/m/myc21/first_paper/ZTF_DESI_data/ZTF_snia_match_DESI_hostgal_Z_filtered.csv"
ztf_df = pd.read_csv(csv_path)

print(f"Loaded {len(ztf_df)} rows from {csv_path}")
print("CSV columns:", ztf_df.columns.tolist()[:10], "...")

# --------------------------------------------------------------------
# 2. Load the FASTSPEC HDU from the FastSpecFit FITS file
# --------------------------------------------------------------------
fastspec_path = "/global/cfs/cdirs/desi/public/dr1/vac/dr1/fastspecfit/iron/v2.1/catalogs/fastspec-iron-main-dark.fits"

with fits.open(fastspec_path) as hdul:
    fastspec_data = hdul["FASTSPEC"].data  # or hdul[1] if unnamed

# Convert to DataFrame (byteswap needed for big-endian FITS arrays)
fastspec_df = pd.DataFrame(np.array(fastspec_data).byteswap().newbyteorder())

print(f"Loaded {len(fastspec_df)} rows from FASTSPEC")

# --------------------------------------------------------------------
# 3. Select only the columns we care about
# --------------------------------------------------------------------
host_cols = ["TARGETID", "VDISP", "LOGMSTAR", "DN4000", "SFR"]
fastspec_subset = fastspec_df[host_cols]

# --------------------------------------------------------------------
# 4. Merge the datasets on TARGETID
# --------------------------------------------------------------------
merged_df = pd.merge(
    ztf_df,
    fastspec_subset,
    on="TARGETID",
    how="left",       # keep all your SN rows, fill NaN if no match
    validate="m:1"    # many SNe to one host
)

print(f"After merge: {len(merged_df)} rows")
print(merged_df[["TARGETID", "LOGMSTAR", "SFR", "VDISP", "DN4000"]].head())

# --------------------------------------------------------------------
# 5. Save the merged table
# --------------------------------------------------------------------
out_path = "/global/homes/m/myc21/first_paper/ZTF_DESI_data/ZTF_snia_match_with_FASTSPEC_hosts.csv"
merged_df.to_csv(out_path, index=False)
print("✅ Saved merged table to:", out_path)
```

```python
import time
from astropy.io import fits
from astropy.table import Table

# --------------------------------------------------------------------
# Paths
# --------------------------------------------------------------------
fits_path = "/global/cfs/cdirs/desi/public/dr1/vac/dr1/fastspecfit/iron/v2.1/catalogs/fastspec-iron-main-dark.fits"
output_path = "ZTF_DESI_DATA/desi_hostprops_fastspec_dark.fits"

start = time.time()

# --------------------------------------------------------------------
# 1. Load only the needed HDU and columns (memory-mapped)
# --------------------------------------------------------------------
with fits.open(fits_path, memmap=True) as hdul:
    fastspec = Table(hdul["FASTSPEC"].data)   # or hdul[1] if unnamed

    # Keep only galaxies (optional but recommended)
    galaxy_rows = fastspec[fastspec["SPECTYPE"] == "GALAXY"]

    # Trim to columns of interest
    keep_cols = [
        "TARGETID", "SPECTYPE", "Z", "ZWARN",
        "VDISP", "VDISP_IVAR",
        "LOGMSTAR",
        "SFR",
        "DN4000", "DN4000_IVAR"
    ]

    trimmed = galaxy_rows[keep_cols]

# --------------------------------------------------------------------
# 2. Save the trimmed table to a new FITS file
# --------------------------------------------------------------------
trimmed.write(output_path, overwrite=True)

end = time.time()
print(f"\n⏳ Total time: {end - start:.2f} seconds")

# --------------------------------------------------------------------
# 3. Quick summary of what you saved
# --------------------------------------------------------------------
with fits.open(output_path) as hdul_out:
    print(f"\n📁 Output file: {output_path}")
    print("📋 Columns:", hdul_out[1].columns.names)
    print("🧾 First few rows:")
    print(Table(hdul_out[1].data[:5]))
```
