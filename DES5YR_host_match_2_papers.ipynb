# Converting Table B1 of Wiseman2020 to a true CSV Value #

To Source Data from:

**Wiseman, P. *et al.* (2020).** *Supernova host galaxies in the Dark Energy Survey: I. Deep coadds, photometry, and stellar masses.*  
*Monthly Notices of the Royal Astronomical Society, 495*(4), 4040–4060.  
[https://doi.org/10.1093/mnras/staa1302](https://doi.org/10.1093/mnras/staa1302)

We look at the following webiste:

https://academic.oup.com/mnras/article/495/4/4040/5840549#supplementary-data

The relevant data table to our reserahc is Table B1 which lists host galaxy properties for each SNIa. Although it is advertised to be in a .csv table, it is actually in a space delimted table so the first step is to convert this pseudo csv file into a true .csv file. 

```python
import pandas as pd

# Path to your space-delimited file
file_path = "/Users/pittsburghgraduatestudent/repos/first_paper/hostgalaxy/Wisemna_2020_Table_B1.csv"

# Read it as a space-delimited file (pandas handles multiple spaces automatically)
df = pd.read_csv(file_path, delim_whitespace=True)

# Save it back as a proper CSV
df.to_csv("/Users/pittsburghgraduatestudent/repos/first_paper/hostgalaxy/Wisemna_2020_Table_B1_converted.csv", index=False)

print("✅ Converted and saved as CSV!")
```
