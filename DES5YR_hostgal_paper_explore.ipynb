```python
import pandas as pd
from astropy.table import Table

# Load the SNIa metadata CSV (contains 'CID')
meta_path = '/Users/pittsburghgraduatestudent/repos/first_paper/DES-SN5YR-1.2/4_DISTANCES_COVMAT/DES-SN5YR_HD+MetaData.csv'
meta_df = pd.read_csv(meta_path)

# Load the host galaxy FITS table (contains 'SNID')
head_path = '/Users/pittsburghgraduatestudent/repos/first_paper/DES-SN5YR-1.2/0_DATA/DES-SN5YR_DES/DES-SN5YR_DES_HEAD.FITS.gz'
host_table = Table.read(head_path, hdu=1)
host_df = host_table.to_pandas()

print("Post conversion DataFrame:")
print(host_df.head(10))

# Check column names for matching key
print('MetaData columns:', meta_df.columns)
print('Host FITS columns:', host_df.columns)

# Try to join on 'CID' (CSV) and 'SNID' (FITS)
# Ensure both are strings for matching
meta_df['CID'] = meta_df['CID'].astype(str)

print(meta_df['CID'].head())

host_df['SNID'] = host_df['SNID'].astype(str)

# Merge the two DataFrames on the SN/CID identifier
merged_df = pd.merge(meta_df, host_df, left_on='CID', right_on='SNID', how='left')



# Display the first few rows of the merged DataFrame
print(merged_df.head())

# Now you can analyze host galaxy properties alongside SNIa metadata!
merged_df
```
