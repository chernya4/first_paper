# The Pantheon+ PV.csv File Has a Lot of SNIa information and Its Associated Galaxy Information Lets See if We Can Fill in The Gap for the Information We Need #

```python
import pandas as pd

# Load the SNe with missing host coordinates
missing_path = "Pantehon_+pv_DES5YR_HostGal_data/snia_with_missing_host_coords.csv"
missing_df = pd.read_csv(missing_path)
print(f"Loaded {len(missing_df)} SNe with missing host coordinates.")

# Load the PV host data
pv_path = "Pantehon_+pv_DES5YR_HostGal_data/all_redshifts_PVs.csv"
pv_df = pd.read_csv(pv_path)

# Merge on SNID to keep only matches, and combine the data
# Using how='inner' ensures only SNIDs present in both DataFrames are retained
merged_df = pd.merge(
    missing_df,
    pv_df,
    on='SNID',
    how='inner',
    suffixes=('_missing', '_pv')  # distinguish overlapping column names
)

# Find unmatched SNIDs (i.e., ones in `missing_df` not present in `pv_df`)
unmatched_df = missing_df[~missing_df['SNID'].isin(pv_df['SNID'])]
print(len(unmatched_df), "SNIDs from the missing hosts were not found in the PV data.")
print(unmatched_df['SNID'])
print(len(merged_df), "SNIDs from the missing hosts were successfully matched with the PV data.")

# Save results
merged_out = "Pantehon_+pv_DES5YR_HostGal_data/snia_missing_hosts_joined_with_PV.csv"
unmatched_out = "Pantehon_+pv_DES5YR_HostGal_data/snia_missing_hosts_not_in_PV.csv"

merged_df.to_csv(merged_out, index=False)
unmatched_df.to_csv(unmatched_out, index=False)

print(f"🔗 Joined matched rows saved to: {merged_out}")
print(f"❌ Unmatched SNIDs saved to: {unmatched_out}")
```

# Turns out we Can Also Look at the IAUC  to find these SNIa #

```python
import pandas as pd

# Load the unmatched SNIa data
unmatched_path = "Pantehon_+pv_DES5YR_HostGal_data/snia_missing_hosts_not_in_PV.csv"
unmatched_df = pd.read_csv(unmatched_path)

# Load the PV data
pv_path = "Pantehon_+pv_DES5YR_HostGal_data/all_redshifts_PVs.csv"
pv_df = pd.read_csv(pv_path)

# Merge SNID from unmatched file with IAUC from PV file
merged_df = pd.merge(
    unmatched_df,
    pv_df,
    left_on='SNID',
    right_on='IAUC',
    how='inner',
    suffixes=('_missing', '_pv')
)

# Find unmatched SNIDs (those that didn't match IAUCs)
still_unmatched_df = unmatched_df[~unmatched_df['SNID'].isin(pv_df['IAUC'])]

# Save results
matched_out = "Pantehon_+pv_DES5YR_HostGal_data/snia_missing_hosts_matched_to_IAUC.csv"
unmatched_out = "Pantehon_+pv_DES5YR_HostGal_data/snia_missing_hosts_not_in_IAUC.csv"

merged_df.to_csv(matched_out, index=False)
still_unmatched_df.to_csv(unmatched_out, index=False)

print(f"✅ Merged SNID ↔ IAUC rows saved to: {matched_out}")
print(f"❌ Still unmatched SNIDs saved to: {unmatched_out}")
```
